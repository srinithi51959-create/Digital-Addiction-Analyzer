```python
import tkinter as tk
from tkinter import *
import random
import pandas as pd

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

# =========================
# LOAD DATASET
# =========================

df = pd.read_csv("dataset done1.csv")

print("Dataset Loaded Successfully!")


# =========================
# ENCODING
# =========================

gender_encoder = LabelEncoder()
eye_encoder = LabelEncoder()
addiction_encoder = LabelEncoder()
status_encoder = LabelEncoder()


df["gender"] = gender_encoder.fit_transform(df["gender"])
df["eye_strain_level"] = eye_encoder.fit_transform(df["eye_strain_level"])
df["addiction_score"] = addiction_encoder.fit_transform(df["addiction_score"])

df["user_improvement_status"] = status_encoder.fit_transform(
    df["user_improvement_status"]
)


print("Encoding Completed!")


# =========================
# FEATURES
# =========================

X = df[
[
    "age",
    "gender",
    "screen_time_limit_hours",
    "total_screen_time_hours",
    "social_media_usage_hours",
    "gaming_usage_hours",
    "study_hours",
    "sleep_hours"
]
]


y = df["user_improvement_status"]



# =========================
# TRAIN MODEL
# =========================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)


model = LogisticRegression(max_iter=1000)

model.fit(X_train, y_train)


print("Model Trained Successfully!")



# =========================
# GUI WINDOW
# =========================

root = Tk()

root.title("Digital Addiction Analyzer")

root.geometry("1100x850")

root.configure(bg="#EAF4FC")



# =========================
# HEADING
# =========================

heading = Label(
    root,
    text="DIGITAL ADDICTION ANALYZER",
    font=("Arial",22,"bold"),
    bg="#EAF4FC",
    fg="navy"
)

heading.pack(pady=15)



# =========================
# INPUT AREA
# =========================

input_frame = Frame(
    root,
    bg="#EAF4FC"
)

input_frame.pack()
button_frame = Frame(root, bg="#EAF4FC")
button_frame.pack(pady=10)   


fields = [
    "Age",
    "Gender (0=Male,1=Female)",
    "Screen Time Limit",
    "Total Screen Time",
    "Social Media Hours",
    "Gaming Hours",
    "Study Hours",
    "Sleep Hours"
]


entries = []


for i, field in enumerate(fields):

    Label(
        input_frame,
        text=field,
        font=("Arial",12),
        bg="#EAF4FC"
    ).grid(
        row=i,
        column=0,
        padx=20,
        pady=5,
        sticky="w"
    )


    entry = Entry(
        input_frame,
        width=30,
        font=("Arial",12)
    )


    entry.grid(
        row=i,
        column=1,
        padx=20,
        pady=5
    )


    entries.append(entry)
            

# =========================
# RESULT AREA
# =========================

result_frame = Frame(root, bg="white", bd=2, relief="solid")
result_frame.pack(fill="both", padx=20, pady=15)

result_label = Label(
    result_frame,
    text="",
    font=("Courier New",12),
    bg="white",
    justify=LEFT,
    anchor="nw"
)
result_label.pack(anchor="nw", padx=15, pady=15)
result_label.config(height=16)


# =========================
# PREDICTION FUNCTION
# =========================
def predict():

    age = int(entries[0].get())
    gender = int(entries[1].get())
    screen_limit = int(entries[2].get())
    total_screen = int(entries[3].get())
    social_media = int(entries[4].get())
    gaming = int(entries[5].get())
    study = int(entries[6].get())
    sleep = int(entries[7].get())

    input_data = pd.DataFrame(
        [[
            age,
            gender,
            screen_limit,
            total_screen,
            social_media,
            gaming,
            study,
            sleep
        ]],
        columns=[
            "age",
            "gender",
            "screen_time_limit_hours",
            "total_screen_time_hours",
            "social_media_usage_hours",
            "gaming_usage_hours",
            "study_hours",
            "sleep_hours"
        ]
    )
def predict():

    age = int(entries[0].get())
    gender = int(entries[1].get())
    screen_limit = int(entries[2].get())
    total_screen = int(entries[3].get())
    social_media = int(entries[4].get())
    gaming = int(entries[5].get())
    study = int(entries[6].get())
    sleep = int(entries[7].get())

    # High Addiction
    if total_screen >= 9 or social_media >= 6 or gaming >= 5:
        addiction_status = "High Digital Addiction"
        improvement_status = "Needs Improvement"
        locked_apps = "Social Media Apps and Gaming Apps"
        allowed_apps = "Phone, Messages, Emergency Calls, Learning Apps"
        locked_duration = "1 hour"
        otp = random.randint(100000, 999999)

    # Moderate Addiction
    elif total_screen >= 6 or social_media >= 4 or gaming >= 3:
        addiction_status = "Moderate Digital Addiction"
        improvement_status = "Average Improvement"
        locked_apps = "Social Media Apps and Gaming Apps"
        allowed_apps = "Phone, Messages, Learning Apps"
        locked_duration = "70 Minutes"
        otp = random.randint(100000, 999999)

    # Low Addiction
    else:
        addiction_status = "Low Digital Addiction"
        improvement_status = "Good Improvement"
        locked_apps = "None"
        allowed_apps = "All Apps"
        locked_duration = "0 Minutes"
        otp = "------"

    result_label.config(
        text=f"""============================================================
             DIGITAL ADDICTION ANALYSIS RESULT
============================================================

Addiction Status        : {addiction_status}

Locked Apps             : {locked_apps}

Allowed Apps            : {allowed_apps}

Locked Duration         : {locked_duration}

Unlock OTP              : {otp}

User Improvement Status : {improvement_status}

============================================================
"""
    )

button_frame = Frame(root, bg="#EAF4FC")
button_frame.pack(pady=10)

predict_btn = Button(
    button_frame,
    text="Predict",
    font=("Arial",12,"bold"),
    bg="green",
    fg="white",
    width=15,
    command=predict
)
predict_btn.grid(row=0, column=0, padx=10)


def clear():
    for entry in entries:
        entry.delete(0, END)
    result_label.config(text="")


clear_btn = Button(
    button_frame,
    text="Clear",
    font=("Arial",12,"bold"),
    bg="red",
    fg="white",
    width=15,
    command=clear
)
clear_btn.grid(row=0, column=1, padx=10)


root.mainloop()
```
